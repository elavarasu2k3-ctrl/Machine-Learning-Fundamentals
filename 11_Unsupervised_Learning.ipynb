## What is Unsupervised Learning?
Unsupervised Learning finds structure in data that has **no target/label column** — there is no "correct answer"
to predict. Instead of learning `X -> y`, the algorithm looks for patterns, groupings, or a simpler representation
within `X` alone.

## Supervised vs Unsupervised Learning
| Aspect | Supervised | Unsupervised |
|---|---|---|
| Labels | Required (`y` exists) | Not required (only `X`) |
| Goal | Predict a known target | Discover structure/patterns |
| Example algorithms | Linear/Logistic Regression, Decision Tree, SVM, KNN, Naive Bayes, Random Forest | K-Means, Hierarchical Clustering, PCA, DBSCAN |
| Evaluation | Accuracy, F1, RMSE, R² (compare to ground truth) | Silhouette score, inertia, explained variance (no ground truth to compare against) |
| Business examples | Churn prediction, price prediction, diagnosis | Customer segmentation, anomaly detection, topic discovery |

## Clustering
Grouping data points so that points in the same group are more similar to each other than to points in other
groups, based purely on feature similarity/distance — no labels involved. Covered in depth in Notebook 12 (K-Means).

## Dimensionality Reduction
Techniques (e.g., PCA) that compress a high-dimensional feature set into fewer dimensions while preserving as much
of the original variance/information as possible. Used for visualization, noise reduction, and speeding up
downstream models.

## When is labelled data unavailable?
- **No historical outcome recorded** — e.g., a company has purchase data but never labeled "high value" customers.
- **Labeling is too expensive/slow** — e.g., manually categorizing millions of support tickets.
- **The goal is exploratory** — discovering unknown groups rather than predicting a predefined one.
- **Anomaly/fraud detection** — fraud is rare and evolving; there's rarely a clean, complete labeled set.

## Applications
- **Customer segmentation** — group customers by behavior without predefined categories.
- **Anomaly detection** — flag points that don't fit any cluster (fraud, sensor faults).
- **Topic discovery** — group similar documents without predefined topic labels.
- **Dimensionality reduction** — PCA to reduce feature count before a supervised model, or for visualization.
- **Gene expression analysis** — clustering patients/genes with similar profiles without diagnostic labels.
